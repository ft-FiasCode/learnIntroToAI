## Lab 13: OpenCV
- Lab Task 1 : Read two images and create a fused image using weighted addition (cv2.addWeighted).

```python
#Importing the required libraries 
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Reading the two images 
img1 = cv2.imread("image1.jpg")
img2 = cv2.imread("image2.jpg")

# for displaying the images
img1 = cv2.cvtColor(img1, cv2.COLOR_BGR2RGB)
img2 = cv2.cvtColor(img2, cv2.COLOR_BGR2RGB)

# Resize the images 
img2 = cv2.resize(img2, (img1.shape[1], img1.shape[0]))

# Display the two input images
plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(img1)
plt.title("Image 1")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(img2)
plt.title("Image 2")
plt.axis("off")

plt.show()
```

```python
# Fusion Image
# Weighted addition (image fusion)
alpha = 0.7
beta  = 0.3
gamma = 0

fused = cv2.addWeighted(img1, alpha, img2, beta, gamma)

# Display fused image
plt.figure(figsize=(6,6))
plt.imshow(fused)
plt.title("Fused Image")
plt.axis("off")
plt.show()

# Save fused image
cv2.imwrite("fused_output.jpg", cv2.cvtColor(fused, cv2.COLOR_RGB2BGR))
```

## Task 2 
- Question : 2. Apply different blurring methods on an image and compare.

```python
# Reading the image
img = cv2.imread("image1.jpg")
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Apply different blurring techniques for comparing
avg_blur = cv2.blur(img, (7, 7))                     # Average Blur
gaussian_blur = cv2.GaussianBlur(img, (7, 7), 0)     # Gaussian Blur
median_blur = cv2.medianBlur(img, 7)                 # Median Blur
bilateral_blur = cv2.bilateralFilter(img, 15, 75, 75) # Bilateral Filter
```

```python
# Display all results in a 2x3 grid
plt.figure(figsize=(14, 10))

plt.subplot(2, 3, 1)
plt.imshow(img)
plt.title("Original Image")
plt.axis("off")

plt.subplot(2, 3, 2)
plt.imshow(avg_blur)
plt.title("Average Blur")
plt.axis("off")

plt.subplot(2, 3, 3)
plt.imshow(gaussian_blur)
plt.title("Gaussian Blur")
plt.axis("off")

plt.subplot(2, 3, 4)
plt.imshow(median_blur)
plt.title("Median Blur")
plt.axis("off")

plt.subplot(2, 3, 5)
plt.imshow(bilateral_blur)
plt.title("Bilateral Filter")
plt.axis("off")

plt.tight_layout()
plt.show()







```

## Task 3
- Question: Detect edges using Canny and find contours.

```python
# Reading image 
img = cv2.imread("image1.jpg")
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

#  Canny Edge Detection
edges = cv2.Canny(img_gray, 100, 200)

# Finding contours
contours, hierarchy = cv2.findContours(edges, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

# Draw contours on a copy of the original
contour_img = img_rgb.copy()
cv2.drawContours(contour_img, contours, -1, (255, 0, 0), 2)  # blue contours
```

```python
# Display results
plt.figure(figsize=(14, 6))

plt.subplot(1, 3, 1)
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(edges, cmap='gray')
plt.title("Canny Edges")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(contour_img)
plt.title("Contours Detected")
plt.axis("off")

plt.tight_layout()
plt.show()
```

## Task 4
- Question - Apply bitwise operations between two images.

```python
# Reading both images
img1 = cv2.imread("image1.jpg")
img2 = cv2.imread("image2.jpg")

# Convertion of the images from BGR into RGB 
img1 = cv2.cvtColor(img1, cv2.COLOR_BGR2RGB)
img2 = cv2.cvtColor(img2, cv2.COLOR_BGR2RGB)

# Resize image2 to match image1
img2 = cv2.resize(img2, (img1.shape[1], img1.shape[0]))

# Applying bitwise operations
bitwise_and = cv2.bitwise_and(img1, img2)
bitwise_or  = cv2.bitwise_or(img1, img2)
bitwise_xor = cv2.bitwise_xor(img1, img2)
bitwise_not = cv2.bitwise_not(img1)   # NOT operation is usually applied on a single image
```

```python
# Display all results
plt.figure(figsize=(14, 10))

plt.subplot(2, 3, 1)
plt.imshow(img1)
plt.title("Image 1")
plt.axis("off")

plt.subplot(2, 3, 2)
plt.imshow(img2)
plt.title("Image 2")
plt.axis("off")

plt.subplot(2, 3, 3)
plt.imshow(bitwise_and)
plt.title("Bitwise AND")
plt.axis("off")

plt.subplot(2, 3, 4)
plt.imshow(bitwise_or)
plt.title("Bitwise OR")
plt.axis("off")

plt.subplot(2, 3, 5)
plt.imshow(bitwise_xor)
plt.title("Bitwise XOR")
plt.axis("off")

plt.subplot(2, 3, 6)
plt.imshow(bitwise_not)
plt.title("Bitwise NOT (on Image 1)")
plt.axis("off")

plt.tight_layout()
plt.show()
```
