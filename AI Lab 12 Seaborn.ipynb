## AI LAB 11 - Seaborn
### Lab Task 1 : Basic Visualization
1. Load the tips dataset.
2. Create:
• A scatter plot (total bill vs tip)
• A bar plot (average total bill per day)
3. Add appropriate titles, axis labels, and color palettes.


```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

- Question 1 - Load the tips datase

```python
tips = pd.read_csv("tip.csv")
# Diplaying the first five rows of the dataset
tips.head()
```

- Question 2 
-  Create:
• A scatter plot (total bill vs tip)
• A bar plot (average total bill per day)

## Scatter Plot (Total bill Vs Tip)

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="day",           # color points by day
    palette="viridis"    # color palette
)

plt.title("Total Bill vs Tip")
plt.xlabel("Total Bill ($)")
plt.ylabel("Tip ($)")

plt.show()
```

## Bar Plot 
- Average total bill per day

```python
plt.figure(figsize=(8,5))

sns.barplot(
    data=tips,
    x="day",
    y="total_bill",
    estimator="mean",
    hue = "day",
    palette = "mako",
    legend = False
)

plt.title("Average Total Bill per Day")
plt.xlabel("Day of the Week")
plt.ylabel("Average Total Bill ($)")

plt.show()
```

## Task 2: Statistical Analysis

```python
# Question 1 - Use a boxplot to compare tip distribution by gender.
plt.figure(figsize=(8,5))

sns.boxplot(
    data=tips,
    x="sex",         
    y="tip",         
    palette="Set2",
    hue="sex",
    legend=False
)

plt.title("Tip Distribution by Gender")
plt.xlabel("Gender")
plt.ylabel("Tip Amount ($)")

plt.show()
```

```python
# Question 2 -  Plot a violin plot showing tip distribution across days.
plt.figure(figsize=(8,5))

sns.violinplot(
    data=tips,
    x="day",         
    y="tip",         
    palette="pastel",
    hue="day",
    legend=False 
)

plt.title("Violin Plot: Tip Distribution Across Days")
plt.xlabel("Day of the Week")
plt.ylabel("Tip Amount ($)")

plt.show()
```

## Task 3: Relationship Exploration
- 1. Load the dataset penguins (built-in Seaborn dataset).
- 2. Create a pairplot to explore relationships among numeric variables.
- 3. Use regplot to draw a regression line between two features (e.g., bill_length_mm vs
bill_depth_mm).


```python
# 1. Load the dataset penguins (built-in Seaborn dataset).
penguins = sns.load_dataset("penguins")
penguins.head()
```

```python
# 2. Create a pairplot to explore relationships among numeric variables.
sns.pairplot(
    penguins.dropna(),
    hue="species",
    palette="Set2",
    height=2
)
plt.show()
```

```python
# 3. Use regplot to draw a regression line between two features (e.g., bill_length_mm vs
# bill_depth_mm).
plt.figure(figsize=(7,5))

sns.regplot(
    data=penguins,
    x="bill_length_mm",
    y="bill_depth_mm",
    color="skyblue"  
)

plt.title("Regression Plot: Bill Length vs Bill Depth")
plt.xlabel("Bill Length (mm)")
plt.ylabel("Bill Depth (mm)")

plt.show()

```
