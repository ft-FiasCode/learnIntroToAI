## AI Lab 09: Pandas
Task 1: Data Cleaning
Dataset: create dataset students.csv (columns: Name, Age, Department, Marks)
Instructions:
1. Load the dataset into Pandas.
2. Identify and fill missing values using mean/median.
3. Remove any duplicate rows.
4. Rename the Marks column to Score.
5. Save the cleaned data as cleaned_students.csv.

```python
## Importing the Pandas Library
import pandas as pd
```

```python
# loading the dataset named student.csv
df = pd.read_csv("students.csv")
df
```

```python
# Identifying and fill the missing values
df.isnull().sum()
```

```python
df["Age"] = df["Age"].fillna(df["Age"].mean())
# As the age is continues so we use the mean to fill in the age
df
```

```python
# Removing duplicate rows
# First checking the duplicate rows
df.duplicated().sum()
```

```python
# As the duplicate exists so removing them
df = df.drop_duplicates()
df
```

```python
df = df.rename(columns={"Marks" : "Score"})
df
```

```python
# Save cleaned dataset
df.to_csv("cleaned_students.csv", index=False)
```

Task 2: Statistical Analysis
Using the cleaned dataset:
1. Find mean, median, max, min, and standard deviation of Score.
2. Group students by Department and calculate average score per department.
3. Sort students by Score (descending order).


```python
# Finding Mean, Median, Max, Min and Standard Deviation(SD) of Score
import pandas as pd
df = pd.read_csv("cleaned_students.csv")

mean_score = df["Score"].mean()
median_score = df["Score"].median()
max_score = df["Score"].max()
min_score = df["Score"].min()
std_score = df["Score"].std()

print("The mean of the score is: ",mean_score)
print("The median of the score is:" ,median_score)
print("The max of the score is: ", max_score)
print("The min of the score is:" ,min_score)
print("The std of the score is:" ,std_score)


```

```python
# Group by department and calculate average score
df.groupby("Department")["Score"].mean()
```

```python
# Sort Students by their score in descending order
df_sorted = df.sort_values(by="Score", ascending = False)
df_sorted
```
