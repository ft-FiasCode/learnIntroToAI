## Task 2: Sensor Data Reshaping (IoT Scenario)
An IoT sensor records temperature every hour for 24 hours:
np.random.randint(18, 40, size=24)

1. Generate the array.
2. Reshape it into a 6×4 array (6 time-blocks, each containing 4 readings).
3. Compute column-wise maximum temperature to identify peak hours.
4. Find all temperature readings above 35°C.

```python
# Genearting the array
import numpy as np
temps = np.random.randint(18, 40, size=24)
print("Hourly Temperatures:",temps)
```

## Task 3: Image Pixel Processing (AI/Computer Vision Scenario)
A simple grayscale image can be represented as a 4×4 matrix of pixel values (0–255):
img = np.random.randint(0, 256, size=(4,4))
Requirements:
1. Generate the image matrix.
2. Flatten the matrix into a 1D array (simulate image preprocessing).
3. Find:
• Minimum pixel value
• Maximum pixel value
• Average brightness
4. Create a new array where all pixels greater than 200 are replaced with 255 (highlighting
bright spots).

```python
import numpy as np

img = np.random.randint(0, 256, size=(4,4))
print("Orginal Image Matrix:\n", img)
```

```python
# Flatten the matrix into 1D
flat_img = img.flatten()
print("\nFlattened Image Matrix: \n", flat_img)
```

```python
# Finding min, max and avg brightness
min_pixel = np.min(img)
max_pixel = np.max(img)
avg_bright = np.mean(img)

print("The min pixel is: ",min_pixel)
print("The max pixel is: ",max_pixel)
print("The average brightness is: ",avg_bright)
```

```python
# Highlightining the bright spots
highlighted = img.copy()
highlighted[highlighted > 200] = 255
print("\nHighlighted Image: \n",highlighted)
```

## Task 4: Sales Dataset Analysis
A small shop tracks daily sales (in rupees) for 10 days:
[1200, 1500, 1100, 980, 1600, 1750, 1450, 1300, 2000, 1850]
Requirements:
1. Convert to NumPy array.
2. Find:
• Total sales
• Average sales
• Day with minimum sales
3. Increase all sales by 10% (festival season effect).
4. Show sales above 1500 after increment.

```python
# Converting into NumPy array
import numpy as np
sales_list = [1200,1500,1100,980,1600,1750,1450,1300,2000,1850]
sales = np.array(sales_list)
sales
```

```python
# Finding Total Average, and day with min sales
total_sales = np.sum(sales)
average_sales = np.mean(sales)
min_day = np.argmin(sales)+1
min_value = np.min(sales)

print("The total sales are: ",total_sales)
print("The Average sales are: ",average_sales)
print(f"The day with minimum sales was {min_day} and the minimum sales were {min_value}")
```

```python
# Incresing the sales by 10% (Season Effect)
increased_sales = sales * 1.10
increased_sales
```

```python
# Showing the sales above 1500 after increment
high_sales = increased_sales[increased_sales > 1500]
print(f"The increased sales above 1500 are {high_sales} ")
```

## Task 5: Scientific Data – Trigonometric Wave
A physics experiment records 15 evenly spaced angles between 0 and 2π.
Requirements:
1. Generate the angles using linspace.
2. Compute:
• Sine values
• Cosine values
3. Find all angles where sin(angle) > 0.8.
4. Plot the sine wave using matplotlib.

```python
import matplotlib.pyplot as plt
```

```python
# Generating 15 angles using linespace
angles = np.linspace(0,2 * np.pi, 15)
print("The angles are:\n",angles)
```

```python
sine_values = np.sin(angles)
cosine_values = np.cos(angles)
print("The sine values are:\n",sine_values)
print("The cosine values are:\n",cosine_values)
```

```python
high_sine_angles = angles[sine_values > 0.8]
print("The High Sine values are: \n", high_sine_angles)
```

```python
# Plot the sine wave using matplotlib
plt.figure(figsize=(6,4))
plt.plot(angles, sine_values)
plt.title("Sine Wave (0 to 2π)")
plt.xlabel("Angle (radians)")
plt.ylabel("sin(angle)")
plt.grid(True)
plt.show()
```
