Task 1: Line and Bar Charts
Create a program to:
1. Plot student marks using a bar chart.
2. Plot two different subjects (Math and Science) on the same line plot with labels and
legends.
3. Add grid lines and customize colors.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("students.csv")
df
```

```python
# Bar chart of Students Marks 
plt.figure(figsize=(10,5))

plt.bar(df["Name"], df["Math"], color = 'skyblue')


plt.title("Math Marks of Students")
plt.xlabel("Students")
plt.ylabel("Marks")

plt.grid(axis='y',linestyle='--', alpha=0.7)


plt.show
```

```python
# Plotting the students marks using the line plot
plt.figure(figsize=(10,5))

plt.plot(df["Name"], df["Math"], marker='o', label="Math")
plt.plot(df["Name"], df["Science"], marker='s', label="Science")

plt.title("Math vs Science Marks")
plt.xlabel("Students")
plt.ylabel("Marks")

plt.grid(True,linestyle='--', alpha=0.6)
plt.legend()

plt.show

```

```python
# Custom Colors and Gridlines
plt.figure(figsize=(10,5))

plt.plot(df["Name"], df["Math"], color='red', linewidth=2, marker='o', label="Math")
plt.plot(df["Name"], df["Science"], color='green', linewidth=2, marker='s', label="Science")

plt.title("Customized Line Plot")
plt.xlabel("Students")
plt.ylabel("Marks")

plt.grid(color='gray', linestyle='--', linewidth=0.5)
plt.legend()

plt.show()
```

Task 2: Statistical Visualization
Using sample data of student ages or salaries:
1. Draw a histogram to show the distribution.
2. Plot a pie chart showing the percentage of students in different departments (CS, IT, AI,
EE).
3. Label all parts clearly with titles and legends.

```python
# Histogram to show the distribution
plt.figure(figsize=(8,5))
plt.hist(df["Age"], bins=5, color='lightgreen',edgecolor = 'black')

plt.title("Distribution of Student Age")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.grid(axis='y', linestyle='--', alpha = 0.6)

plt.show()
```

```python
# Pie chart to show percentage of students by Deptt
dept_counts = df["Department"].value_counts()
dept_counts
```

```python
plt.figure(figsize=(7,7))

plt.pie(dept_counts, labels=dept_counts.index, autopct='%1.1f%%',
       startangle=90, shadow=True)
plt.title("Percentage of Students in Each Department")
plt.legend(title="Department")
plt.show()
```

Task 3: Scatter and Subplots
1. Create a scatter plot for height vs weight data.
2. Create two subplots — one for a line graph and one for a bar chart.
3. Give proper titles and axis labels for both.

```python
# Scatter plot for Height vs Weight
plt.figure(figsize=(8,5))

plt.scatter(df["Height"], df["Weight"], color='red', marker = 'o')

plt.title("Height vs Weight")
plt.xlabel("Height(cm)")
plt.ylabel("Weight(kg)")
plt.grid(True, linestyle= '--', alpha = 0.5)

plt.show()
```

```python
# Two Subplots - One Line Graph + One Bar Chart
plt.figure(figsize=(12,5))

#  Subplot 1: Line Graph
plt.subplot(1, 2, 1)
plt.plot(df["Name"], df["Math"], marker='o', color='blue')
plt.title("Math Marks Line Graph")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.grid(True, linestyle='--', alpha=0.5)

#  Subplot 2: Bar Chart 
plt.subplot(1, 2, 2)
plt.bar(df["Name"], df["Science"], color='green')
plt.title("Science Marks Bar Chart")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.tight_layout()  # adjusts spacing so labels don’t overlap
plt.show()
```
